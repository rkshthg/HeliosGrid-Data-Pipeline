```python
from pyspark.sql import functions as F
```

```python
# 1. Ingest Forecast Telemetry -> Bronze Delta Table
(spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", "/Volumes/heliosgrid_catalog/bronze/landing_raw_json/checkpoints/forecast/schema")
    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
    .option("cloudFiles.rescuedDataColumn", "_rescued_data")
    .load("/Volumes/heliosgrid_catalog/bronze/landing_raw_json/forecast/")
    .withColumn("_ingested_timestamp", F.current_timestamp())
    .withColumn("_source_file", F.expr("_metadata.file_path"))
    .writeStream
    .format("delta")
    .option("checkpointLocation", "/Volumes/heliosgrid_catalog/bronze/landing_raw_json/checkpoints/forecast/write")
    .outputMode("append")
    .trigger(availableNow=True)
    .table("heliosgrid_catalog.bronze.raw_forecast_telemetry")
)
```

```python
# 2. Ingest Air Quality Telemetry -> Bronze Delta Table
(spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", "/Volumes/heliosgrid_catalog/bronze/landing_raw_json/checkpoints/air_quality/schema")
    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
    .option("cloudFiles.rescuedDataColumn", "_rescued_data")
    .load("/Volumes/heliosgrid_catalog/bronze/landing_raw_json/air_quality/")
    .withColumn("_ingested_timestamp", F.current_timestamp())
    .withColumn("_source_file", F.expr("_metadata.file_path"))
    .writeStream
    .format("delta")
    .option("checkpointLocation", "/Volumes/heliosgrid_catalog/bronze/landing_raw_json/checkpoints/air_quality/write")
    .outputMode("append")
    .trigger(availableNow=True)
    .table("heliosgrid_catalog.bronze.raw_air_quality_telemetry")
)
```
